```python
#juego_escalera_culebras
import numpy as np
import matplotlib.pyplot as plt
N=101
P=np.eye(N)+0.0
saltos=[(1,38),(4,14),(8,30),(21,42),(28,76),(32,10),(36,6),(48,26),(50,67),(62,17),(71,92),(81,99),(95,44),(97,78)]
salto={}
for org, dst in saltos:
    salto[org]=dst



for i in range(N):
    if(i not in salto.keys()):
        for j in range(i+1, i+6+1):
            if(j in salto.keys()):
                P[i,salto[j]]+=1.0/6
                P[i,i]-=1.0/6
            else:
                if(j<N):
                    P[i,j]+=1.0/6
                    P[i,i]-=1.0/6
P[-1,-1]=1
Pacum=np.cumsum(P,axis=1)
plt.imshow(P)
plt.title('Matriz de probabilidades de transición')
plt.colorbar()

def nextState(pr):
    return np.min(np.where(pr>=np.random.rand()))

tr=[]
Lmin=1e3
for i in range(5000):
    s=[0]
    while(s[-1]!=100):
        sn=nextState(Pacum[s[-1]])
        s.append(int(sn))
    if(len(s)<Lmin):
        Lmin=len(s)
        smin=s.copy()
        print('Número mínimo de saltos: ',Lmin, '. Secuencia: ', s)
    tr.append(s)
    #print(i)
```
