# Modelo markoviano inicial

Comprendamos los conceptos fundamentales

```python
import numpy as np
```

```python


np.set_printoptions(precision=6, suppress=True)

# --------------------------------------------------
# 1. Matriz de transición
#    p_ij = P(X_{t+1}=j | X_t=i)
#    Las filas deben sumar 1
# --------------------------------------------------

P = np.array([
    [0.6, 0.3, 0.1],
    [0.2, 0.5, 0.3],
    [0.1, 0.3, 0.6]
])

print("Matriz de transición P:")
print(P)

print("\nSuma de las filas:")
print(P.sum(axis=1))

```

```python

# --------------------------------------------------
# 2. Distribución inicial
#    Vector fila
# --------------------------------------------------


# Comenzamos con certeza en el estado 3
pi0 = np.array([1.0, 0.0, 0.0])

print("\nDistribución inicial:")
print(pi0)


# --------------------------------------------------
# 3. Evolución de la distribución
#
#    pi_(t+1) = pi_t P
# --------------------------------------------------

pi = pi0.copy()

print("\nEvolución de la distribución:")

for t in range(20):
    print(f"t={t:2d}: {pi}")
    pi = pi @ P

```

```python

# --------------------------------------------------
# 4. Potencias de P
#
#    (P^n)_ij =
#    P(X_{t+n}=j | X_t=i)
# --------------------------------------------------

for n in [1, 2, 5, 10, 50, 100]:

    Pn = np.linalg.matrix_power(P, n)

    print(f"\nP^{n}:")
    print(Pn)


```

```python
# --------------------------------------------------
# 5. Distribución estacionaria por punto fijo
#
#    Iteramos:
#
#       pi_(t+1) = pi_t P
#
#    hasta que:
#
#       pi_(t+1) ≈ pi_t
# --------------------------------------------------

pi = pi0.copy()

tolerancia = 1e-10
max_iteraciones = 10000

for k in range(max_iteraciones):

    pi_nueva = pi @ P

    error = np.max(np.abs(pi_nueva - pi))

    if error < tolerancia:
        break

    pi = pi_nueva


print("\nDistribución estacionaria:")
print(pi_nueva)

print("Iteraciones:", k + 1)
print("Error:", error)


# --------------------------------------------------
# 6. Comprobación del punto fijo
# --------------------------------------------------

print("\nComprobación:")

print("pi:")
print(pi_nueva)

print("pi @ P:")
print(pi_nueva @ P)
```

# Modelo con estados absorbentes

```python
import numpy as np

np.set_printoptions(precision=6, suppress=True)

# Estados:
# 0 = A  transitorio
# 1 = B  transitorio
# 2 = C  absorbente
# 3 = D  absorbente

estados = ["A", "B", "C", "D"]

# --------------------------------------------------
# 1. Matriz de transición
#
# p_ij = P(X_(t+1)=j | X_t=i)
#
# Las filas son el estado de origen
# Las columnas son el estado de destino
# --------------------------------------------------

P = np.array([
    [0.3, 0.3, 0.3, 0.1],   # desde A
    [0.2, 0.3, 0.1, 0.4],   # desde B
    [0.0, 0.0, 1.0, 0.0],   # C absorbente
    [0.0, 0.0, 0.0, 1.0]    # D absorbente
])

print("Matriz de transición P:")
print(P)

print("\nSuma de las filas:")
print(P.sum(axis=1))

```

```python

# --------------------------------------------------
# 2. Distribución inicial
#
# Comenzamos con certeza en B
# --------------------------------------------------

pi0 = np.array([0.0, 1.0, 0.0, 0.0])

print("\nDistribución inicial:")
print(pi0)


# --------------------------------------------------
# 3. Evolución de la cadena
#
# pi_(t+1) = pi_t P
# --------------------------------------------------

pi = pi0.copy()

print("\nEvolución:")

for t in range(30):

    print(
        f"t={t:2d}  "
        f"A={pi[0]:.4f}  "
        f"B={pi[1]:.4f}  "
        f"C={pi[2]:.4f}  "
        f"D={pi[3]:.4f}"
    )

    pi = pi @ P

```

```python

# --------------------------------------------------
# 4. Potencias de P
# --------------------------------------------------

for n in [1, 2, 5, 10, 50]:

    Pn = np.linalg.matrix_power(P, n)

    print(f"\nP^{n}:")
    print(Pn)

```

```python

# --------------------------------------------------
# 5. Convergencia por punto fijo
# --------------------------------------------------

pi = pi0.copy()

tolerancia = 1e-10
max_iteraciones = 10000

for k in range(max_iteraciones):

    pi_nueva = pi @ P

    error = np.max(np.abs(pi_nueva - pi))

    if error < tolerancia:
        break

    pi = pi_nueva


print("\nDistribución final:")
print(pi_nueva)

print("Iteraciones:", k + 1)
print("Error:", error)


# --------------------------------------------------
# 6. Comprobación
# --------------------------------------------------

print("\npi_final @ P:")
print(pi_nueva @ P)
```
